```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
import urllib.request

```

```python
urllib.request.urlretrieve('https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg', 'input.jpg')

img = cv2.imread('input.jpg')
x, y, w, h = 100, 100, 150, 150
cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 3)
extracted = img[y:y+h, x:x+w]
cv2_imshow(img)
cv2_imshow(extracted)
```
